```python
class Car:
    def __init__(self, *args):              # Dynamic parameter
        if len(args) == 1:
            self.name = args[0]
        elif len(args) == 2:
            self.name = args[0]
            self.engine = args[1]           # This is method use to print a specific parameter, No need to give all parameter values.
        elif len(args) == 3:
            self.name = args[0]
            self.engine = args[1]
            self.speed = args[2]

    def electric_vehicle(self):
        print("Hello this {} Electric vehicle with {} engine and Top Speed is {}.".format(self.name,self.engine, self.speed))
```

```python
car1 = Car("BMW","m3000", "200km/h")
print(car1.name)
print(car1.engine)
print(car1.speed)
car1.electric_vehicle()
```

# Inheritance in Python

```python
# make Car class blueprint
class Car:
    def __init__(self,seat,engine,name):
        self.seat = seat
        self.engine = engine
        self.name = name\

    def car_driving(self):
            print("This is car Driving function")
            
```

```python
class BMW(Car):    # we inherite the Car class
    def __init__(self,seat,engine,name,speed,price):
        super().__init__(seat,engine,name)    # assign the inherite parameter
        self.speed = speed
        self.price = price

bmw_n3 = BMW(4,"v8","BMW","230km/h",15000000)
car1 = Car(2,"v8","BMW")
print("Hello this is {} Car with {} engine and top speed {}".format(bmw_n3.name,bmw_n3.engine,bmw_n3.speed))
```

```python
print(type(car1))
print(type(bmw_n3))
```

```python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display(self):
        print(f"This is Car make {self.make}, Car Model {self.model} and Release Year {self.year}.")

    def start_engine(self):
        print(self.make,"Engine Start of ")
```
