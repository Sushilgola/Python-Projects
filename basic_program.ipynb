```python
# Reverse the number
num = int(input("Number: "))
rev = 0
while num != 0:
    rem = num%10
    rev = rev*10 + rem
    num //= 10
print(rev)
    
    
    
```

```python
# palindrome string
string = input("Enter String: ")
l_string = string.lower()
is_palindrome = True
j = -1
for i in range(round((len(l_string)/2))):
    if l_string[i] == l_string[j]:
        is_palindrome = True
        j -= 1
    else:
        is_palindrome = False
if is_palindrome:
    print("Palindrome")
else:
    print("Not A Palindrome")
```

```python
# Count vowel and conconent in string
string = input("Enter String: ")
vowel = 0
consonent = 0
for i in string:
    if i=='a' or i=='e' or i=='i' or i=='o' or i=='u':
        vowel += 1
    else:
        consonent += 1
print("Vowels: ",vowel)
print("Consonent: ", consonent)
```

```python
# remove a specific element in the list
lst = [12,23,19,10,17,20,28,24,27,28,30,18,15,14,32,35]
num = int(input("Enter the number to remove frome list: "))
if num not in lst:
    print("Number is not in list")
for i in lst:
    if i == num:
        lst.remove(i)
print(lst)
```

```python
# find the second largest element in list
lst = [12,23,19,10,17,20,28,24,27,28,30,18,15,14,38,35]
largest = 0
s_largest = 0
for i in range(len(lst)):
    if lst[i] > largest:
        largest = lst[i]
for i in lst:
    if i > s_largest and i < largest:
        s_largest = i
print("Second largest number: ", s_largest)
```

```python
#find maximum number
# size = int(input("Enter Array Size: "))
arr = list(map(int, input("Enter Number :").split()))
maximum = 0
for n in range(len(arr)):
    if arr[n] > maximum:
        maximum = arr[n]
print("Max Number", maximum)

```

```python
# find second largest number
arr = list(map(int, input("Enter Number :").split()))
largest = -2*31
second_largest = -2*31
for x in arr:
    if x > largest:
        largest = x
for n in arr:
    if n > second_largest and n < largest:
        second_largest = n
print(second_largest)
```
