## OOPS In python
### Child Class Acess All properties of parent class

```python
class Student:
    def __init__(self, name, rollno):
        self.name = name
        self.rollno = rollno
    def stu_details(self):
        print(f"Name {self.name} and His Roll No {self.rollno}.")

class Student1(Student):
    def __init__(self, name, rollno, course):
        super().__init__(name, rollno)
        self.course = course
    def print_detail(self):
        
        print(f"Name {self.name} , Roll No {self.rollno} and Course {self.course}.")
        
```

```python
stu1 = Student1("Sushil", 146, "Mca")
stu1.stu_details()
stu1.print_detail()
```

## Polymorphism In Python
### 1 CompileTime Polymorphism (method overloading)  2  RunTime Polymorphism (method overriding)
#### 1. Same name function but different type of Parameter
#### 2. same name function but different class methods

```python
#  function Overloading

class Calculator:
    def sum_num(self, num1, num2=0, num3=0):
        return num1+num2+num3

cal = Calculator()
print(cal.sum_num(1,2,3))
print(cal.sum_num(2,3))
print(cal.sum_num(4,0,4))
    
```

## 2. Method Overriding

```python
class Animal:
    def sound(self):
        print("Animal Sounds")
class Cat(Animal):
    def sound(self):
        print("Meaw")
class Lion(Animal):
    def sound(self):
        print("Rore")


for an in [Cat(), Lion(), Animal()]:
    an.sound()
```

## Encapsulation in Python 
### To hide the personal data from users.

```python
# Public and Private
class Parent:
    def __init__(self, name, age, salary, crush):
        self.name = name    # Public variable
        self.age = age      # public variable
        self.__salary = salary    # Private variable
        self._crush = crush

    def print_detail(self):
        print("Name: ", self.name, ", Age: ", self.age, ", Salary:",self.__salary, ", Crush:", self._crush)
```

```python
p = Parent("Sushil",21, 1000000, "ABC")
p.print_detail()
print(p.name)
print(p._crush)  # Protected Variable 
# print(p.__salary)   # private variable not direct accesible
```

#### Encapsulation using getter and setter

```python
class Student:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age
    def get_name(self):
        return self.__name
    def set_name(self, name):
        self.__name = name

    def get_age(self):
        return self.__age
    def set_age(self, age):
        if self.__age >0:
            self.__age = age
        else:
            print("You Type Wrong Age")
        

s = Student("Sushil", 21)
print(s.get_name())
print(s.get_age())
s.set_name("Sushil")
s.set_age(-6)
s.set_age(45)
```
